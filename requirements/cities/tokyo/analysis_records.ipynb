```python
import pandas as pd
import matplotlib.pyplot as plt

# Published JMA monthly climate normals (1991–2020)
# Transcribed from the Tokyo city requirements research notes.
monthly_normals = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "mean_tmax_c": [9.8, 10.9, 14.2, 19.4, 23.6, 26.1,
                    29.9, 31.3, 27.5, 22.0, 16.7, 12.0],
    "mean_tmin_c": [1.2, 2.1, 5.0, 9.8, 14.6, 18.5,
                    22.4, 23.5, 20.3, 14.8, 8.8, 3.8],
    "rainfall_mm": [59.7, 56.5, 116.0, 133.7, 139.7, 167.8,
                    156.2, 154.7, 224.9, 234.8, 96.3, 57.9],
    "relative_humidity_pct": [51, 52, 57, 62, 68, 75,
                              76, 74, 75, 71, 64, 56],
    "sunshine_hours": [192.6, 170.4, 175.3, 178.8, 179.6, 124.2,
                       151.4, 174.2, 126.7, 129.4, 149.8, 174.4]
})

monthly_normals
```

## 2. Initial observations from the research

These are preliminary interpretations of the monthly normals and the research already recorded. Revisit them when we analyse actual daily observations.

1. **Seasonality:** Mean maximum temperature rises from 9.8°C in January to 31.3°C in August.
2. **Summer humidity:** Relative humidity is around 74–76% from June through September.
3. **Rainfall:** September and October have the highest monthly normal rainfall in this table.
4. **Winter:** January has the lowest mean maximum and minimum temperatures.
5. **Model implication:** An 8-day temperature-lag baseline may not represent the time of year explicitly. Seasonal features can be tested later as a separate experiment (V1.1), not silently added to the V1 baseline.

### Questions to investigate with daily data

- How does the distribution of daily maximum temperature vary by month?
- How many days reach 35°C or above in each year?
- Are extreme-heat days concentrated in particular months?
- Are there missing observations or station-history discontinuities?


```python
# Plot the monthly mean maximum and minimum temperatures
plt.figure(figsize=(11, 5))
plt.plot(monthly_normals["month"], monthly_normals["mean_tmax_c"],
         marker="o", label="Mean maximum temperature")
plt.plot(monthly_normals["month"], monthly_normals["mean_tmin_c"],
         marker="o", label="Mean minimum temperature")
plt.title("Tokyo: Monthly Mean Temperatures (JMA 1991–2020 Normals)")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Plot monthly rainfall and relative humidity
fig, ax1 = plt.subplots(figsize=(11, 5))

ax1.bar(monthly_normals["month"], monthly_normals["rainfall_mm"], alpha=0.7)
ax1.set_xlabel("Month")
ax1.set_ylabel("Rainfall (mm)")
ax1.set_title("Tokyo: Monthly Rainfall and Relative Humidity (JMA 1991–2020 Normals)")
ax1.grid(axis="y", alpha=0.25)

ax2 = ax1.twinx()
ax2.plot(monthly_normals["month"], monthly_normals["relative_humidity_pct"],
         marker="o", linewidth=2)
ax2.set_ylabel("Relative humidity (%)")

fig.tight_layout()
plt.show()
```

## 3. Initial research conclusions

**Findings recorded:**
- Tokyo has a pronounced annual temperature cycle.
- The warmest monthly normal for mean maximum temperature is August.
- Summer months have higher relative humidity than winter months.
- Rainfall normals are highest in September and October.

**Potential impact on HeatShield:**
- Keep the common V1 input definition unchanged: eight consecutive daily maximum-temperature observations.
- Treat seasonal/calendar variables as candidates for a later controlled experiment.
- Keep humidity and rainfall outside the initial V1 feature set unless the project scope is formally revised.
